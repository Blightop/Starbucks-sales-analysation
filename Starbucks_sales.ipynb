```python
offer_events = transcript_clean[
    transcript_clean["event"].isin(
        ["offer received", "offer viewed", "offer completed"]
    )
].copy()
```

```python
offer_events = offer_events.merge(
    portfolio_clean[["id", "offer_type"]],
    left_on="offer_id",
    right_on="id",
    how="left"
)
```

```python
offer_events = offer_events.drop(columns="id" and "value_dict")
```

```python
display(offer_events.head())
```

```python
received_by_type = (
    offer_events[offer_events["event"] == "offer received"]
    .drop_duplicates(["person", "offer_id"])
    .groupby("offer_type")
    .size()
)

viewed_by_type = (
    offer_events[offer_events["event"] == "offer viewed"]
    .drop_duplicates(["person", "offer_id"])
    .groupby("offer_type")
    .size()
)

view_rate_by_type = (
    viewed_by_type / received_by_type * 100
).reset_index(name="view_rate")

display(view_rate_by_type)
```

```python
plt.figure(figsize=(8, 5))

bar_colors = ['tab:red', 'tab:blue', 'tab:green']

plt.bar(
    view_rate_by_type["offer_type"],
    view_rate_by_type["view_rate"],
    color=bar_colors
    )

plt.ylabel("View rate (%)")
plt.xlabel("Offer type")
plt.title("Offer View Rate by Offer Type")

plt.ylim(0, 100)

for i, rate in enumerate(view_rate_by_type["view_rate"]):
    plt.text(i, rate + 2, f"{rate:.1f}%", ha="center")

plt.show()
```

```python
received_by_type = (
    offer_events[
        (offer_events["event"] == "offer received") &
        (offer_events["offer_type"].isin(["bogo", "discount"]))
    ]
    .drop_duplicates(["person", "offer_id"])
    .groupby("offer_type")
    .size()
)

completed_by_type = (
    offer_events[
        (offer_events["event"] == "offer completed") &
        (offer_events["offer_type"].isin(["bogo", "discount"]))
    ]
    .drop_duplicates(["person", "offer_id"])
    .groupby("offer_type")
    .size()
)

completion_rate_by_type = (
    completed_by_type / received_by_type * 100
).reset_index(name="completion_rate")

display(completion_rate_by_type)
```

```python
plt.figure(figsize=(8, 5))

bar_colors = ['tab:red', 'tab:blue']

plt.bar(
    completion_rate_by_type["offer_type"],
    completion_rate_by_type["completion_rate"],
    color=bar_colors
)

plt.ylabel("Completion rate (%)")
plt.xlabel("Offer type")
plt.title("Offer Completion Rate by Offer Type")

plt.ylim(0, 100)

for i, rate in enumerate(completion_rate_by_type["completion_rate"]):
    plt.text(i, rate + 2, f"{rate:.1f}%", ha="center")

plt.show()
```

```python
offer_performance = (
    offer_events
    .groupby(["offer_id", "offer_type", "event"])
    .size()
    .unstack(fill_value=0)
    .reset_index()
)

display(offer_performance)
```

```python
offer_performance = offer_performance.rename(columns={
    "offer received": "received",
    "offer viewed": "viewed",
    "offer completed": "completed"
})

display(offer_performance)
```

```python
offer_performance = offer_performance.merge(
    portfolio_clean[
        ["id", "reward", "difficulty", "duration", "channels"]
    ],
    left_on="offer_id",
    right_on="id",
    how="left"
)

offer_performance = offer_performance.drop(columns="id")
```

```python
offer_performance["view_rate"] = (
    offer_performance["viewed"] /
    offer_performance["received"] * 100
)

offer_performance["completion_rate"] = (
    offer_performance["completed"] /
    offer_performance["received"] * 100
)
```

```python
offer_performance = offer_performance[
    [
        "offer_id",
        "offer_type",
        "reward",
        "difficulty",
        "duration",
        "channels",
        "received",
        "viewed",
        "completed",
        "view_rate",
        "completion_rate"
    ]
]
```

```python
display(
    offer_performance.sort_values(
        "completion_rate",
        ascending=False
    )
)
```

```python
offer_performance["offer"] = (
    "Offer " +
    (offer_performance.index + 1).astype(str)
)
```

```python
offer_performance = offer_performance[
    [
        "offer",
        "offer_id",
        "offer_type",
        "reward",
        "difficulty",
        "duration",
        "channels",
        "received",
        "viewed",
        "completed",
        "view_rate",
        "completion_rate"
    ]
]

display(offer_performance)
```

```python
plot_data = offer_performance.sort_values(
    "completion_rate",
    ascending=False
)

plt.figure(figsize=(10, 6))

plt.bar(
    plot_data["offer"],
    plot_data["completion_rate"]
)

plt.ylabel("Completion rate (%)")
plt.xlabel("Offer")
plt.title("Completion Rate by Individual Offer")
plt.ylim(0, 100)

for i, rate in enumerate(plot_data["completion_rate"]):
    plt.text(
        i,
        rate + 2,
        f"{rate:.1f}%",
        ha="center"
    )

plt.xticks(rotation=45)

plt.show()
```
